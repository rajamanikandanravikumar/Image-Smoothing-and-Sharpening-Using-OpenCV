```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

```

```python
image = cv2.imread('pomo.png')
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image_rgb)
plt.axis('off')
plt.title("Original Image")
plt.show()
```

```python
avg = cv2.blur(image, (25, 25))
avg_rgb = cv2.cvtColor(avg, cv2.COLOR_BGR2RGB)
plt.imshow(avg_rgb)
plt.title("Averaging Filter")
plt.axis('off')
plt.show()
```

```python
kernel = cv2.getGaussianKernel(3, 0)
kernel = kernel @ kernel.T
```

```python
filtered_image = cv2.filter2D(image, -1, kernel)
filtered_rgb = cv2.cvtColor(filtered_image, cv2.COLOR_BGR2RGB)
plt.imshow(filtered_rgb)
plt.title("Weighted Averaging Filter")
plt.axis('off')
plt.show()
```

```python
g_image = cv2.GaussianBlur(image, (55, 55), 0)
g_rgb = cv2.cvtColor(g_image, cv2.COLOR_BGR2RGB)
plt.imshow(g_rgb)
plt.title("Gaussian Filter")
plt.axis('off')
plt.show()
```

```python
m_image = cv2.medianBlur(image, 25)
m_rgb = cv2.cvtColor(m_image, cv2.COLOR_BGR2RGB)
plt.imshow(m_rgb)
plt.title("Median Filter")
plt.axis('off')
plt.show()
```

```python
sharp_kernel = -kernel.copy()
sharp_kernel[1, 1] = 5
```

```python
sharp = cv2.filter2D(image, -1, sharp_kernel)
sharp_rgb = cv2.cvtColor(sharp, cv2.COLOR_BGR2RGB)
plt.imshow(sharp_rgb)
plt.title("Laplacian Sharpening")
plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
plt.show()
```

```python
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Operator")
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(3, 3, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis('off')

plt.subplot(3, 3, 2)
plt.imshow(avg_rgb)
plt.title("Averaging Filter")
plt.axis('off')

plt.subplot(3, 3, 3)
plt.imshow(filtered_rgb)
plt.title("Weighted Averaging")
plt.axis('off')

plt.subplot(3, 3, 4)
plt.imshow(g_rgb)
plt.title("Gaussian Filter")
plt.axis('off')

plt.subplot(3, 3, 5)
plt.imshow(m_rgb)
plt.title("Median Filter")
plt.axis('off')

plt.subplot(3, 3, 6)
plt.imshow(sharp_rgb)
plt.title("Laplacian Sharpening")
plt.axis('off')

plt.subplot(3, 3, 7)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Operator")
plt.axis('off')

plt.tight_layout()
plt.show()
```
